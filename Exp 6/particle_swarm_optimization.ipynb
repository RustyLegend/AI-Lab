```python
import random
import math
```

```python
def rastrigin(x, y):
    return 20 + (x**2) + (y**2) - 10 * math.cos(2 * math.pi * x) - 10 * math.cos(2 * math.pi * y)
```

```python
def execute_pso(w, c1, c2, num_particles=30, max_iterations=100, bound=5.12):
    particles = []

    for _ in range(num_particles):
        x = random.uniform(-bound, bound)
        y = random.uniform(-bound, bound)
        vx = random.uniform(-1, 1)
        vy = random.uniform(-1, 1)
        fitness = rastrigin(x, y)

        particles.append([(x, y), (vx, vy), (x, y), fitness])

    global_best_pos = None
    global_best_fitness = float('inf')

    for p in particles:
        if p[3] < global_best_fitness:
            global_best_fitness = p[3]
            global_best_pos = p[2]

    for _ in range(max_iterations):

        for p in particles:
            x, y = p[0]
            vx, vy = p[1]
            px, py = p[2]

            r1x, r1y = random.random(), random.random()
            r2x, r2y = random.random(), random.random()

            new_vx = (
                w * vx
                + c1 * r1x * (px - x)
                + c2 * r2x * (global_best_pos[0] - x)
            )

            new_vy = (
                w * vy
                + c1 * r1y * (py - y)
                + c2 * r2y * (global_best_pos[1] - y)
            )

            new_x = x + new_vx
            new_y = y + new_vy

            new_x = max(-bound, min(bound, new_x))
            new_y = max(-bound, min(bound, new_y))

            new_fitness = rastrigin(new_x, new_y)

            if new_fitness < p[3]:
                p[2] = (new_x, new_y)
                p[3] = new_fitness

            p[0] = (new_x, new_y)
            p[1] = (new_vx, new_vy)

        for p in particles:
            if p[3] < global_best_fitness:
                global_best_fitness = p[3]
                global_best_pos = p[2]

    return global_best_fitness
```

```python
def get_mean_fitness(w, c1, c2, runs=5):
    total_fitness = sum(
        execute_pso(w, c1, c2)
        for _ in range(runs)
    )
    return total_fitness / runs
```

```python
print("Objective: Minimize Multimodal Rastrigin Function\n")

print("--- Table 1: Effect of inertia weight w ---")
print(f"{'Case':<15} | {'w':<4} | {'c1':<4} | {'c2':<4} | {'Mean Best Fitness'}")
print("-" * 55)

for case, w in [
    ("Low inertia", 0.4),
    ("Medium inertia", 0.7),
    ("High inertia", 0.9)
]:
    mean_fit = get_mean_fitness(w, 2.0, 2.0)

    print(
        f"{case:<15} | {w:<4} | {2.0:<4} | {2.0:<4} | {mean_fit:.6f}"
    )

print("\n--- Table 2: Effect of cognitive coefficient c1 ---")
print(f"{'Case':<15} | {'w':<4} | {'c1':<4} | {'c2':<4} | {'Mean Best Fitness'}")
print("-" * 55)

for case, c1 in [
    ("Low c1", 0.5),
    ("Medium c1", 1.5),
    ("High c1", 2.5)
]:
    mean_fit = get_mean_fitness(0.7, c1, 2.0)

    print(
        f"{case:<15} | {0.7:<4} | {c1:<4} | {2.0:<4} | {mean_fit:.6f}"
    )


print("\n--- Table 3: Effect of social coefficient c2 ---")
print(f"{'Case':<15} | {'w':<4} | {'c1':<4} | {'c2':<4} | {'Mean Best Fitness'}")
print("-" * 55)

for case, c2 in [
    ("Low c2", 0.5),
    ("Medium c2", 1.5),
    ("High c2", 2.5)
]:
    mean_fit = get_mean_fitness(0.7, 2.0, c2)

    print(
        f"{case:<15} | {0.7:<4} | {2.0:<4} | {c2:<4} | {mean_fit:.6f}"
    )


print("\n--- Table 4: Overall Comparison Table ---")
print(f"{'Case':<15} | {'w':<4} | {'c1':<4} | {'c2':<4} | {'Mean Best Fitness'}")
print("-" * 55)

experiments = [
    ("Baseline", 0.7, 2.0, 2.0),
    ("Low w", 0.4, 2.0, 2.0),
    ("High w", 0.9, 2.0, 2.0),
    ("Low c1", 0.7, 0.5, 2.0),
    ("High c1", 0.7, 2.5, 2.0),
    ("Low c2", 0.7, 2.0, 0.5),
    ("High c2", 0.7, 2.0, 2.5)
]

for case, w, c1, c2 in experiments:
    mean_fit = get_mean_fitness(w, c1, c2)

    print(
        f"{case:<15} | {w:<4} | {c1:<4} | {c2:<4} | {mean_fit:.6f}"
    )


print(
    "\nObservation: Good performance requires balancing exploration "
    "(higher w) and exploitation (lower w). The results may vary "
    "slightly per execution because PSO uses random numbers."
)
```
