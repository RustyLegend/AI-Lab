```python
from collections import deque
```

```python
print('----INPUT----')
v = int(input('Enter no.of vertices: '))
e = int(input('Enter no.of edges: '))

graph = [[0 for i in range(v)] for i in range(v)]

print('Source, Destination pairs: ')
for i in range(e):
  src, dest = map(int, input().split())
  graph[src-1][dest-1] = 1
  graph[dest-1][src-1] = 1

print('Enter source vertex: ')
src = int(input()) - 1

print('Enter goal vertex: ')
goal = int(input()) - 1
```

```python
print(graph)
```

```python
visited = [False for i in range(v)]
q = deque()
q.append(src)
visited[src] = True
parent = [-1]*v
found = 0
```

```python
expansion_path = []
while q:
  curr = q.popleft()
  expansion_path.append(curr)

  if curr == goal:
    found = 1

  for i in range(v):
    if graph[curr][i] != 0 and not visited[i]:
      visited[i] = True
      parent[i] = curr
      q.append(i)

print('\nExpansion Order: ')
for x in expansion_path:
  print(x + 1, end = ' -> ')

print('\n\nShortest Path: ')
shortest_path = []
if found:
  i = goal
  while i != -1:
    shortest_path.append(i)
    i = parent[i]
  for x in reversed(shortest_path):
    print(x + 1, end = ' -> ')
else:
  print('No path found')
```
