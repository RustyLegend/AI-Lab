```python
from heapq import heappush, heappop

print("----INPUT----")

v = int(input("Enter no.of vertices: "))
e = int(input("Enter no.of edges: "))

graph = [[0]*v for _ in range(v)]

print("Enter Source Destination Weight:")
for _ in range(e):
    u, w, cost = map(int, input().split())
    graph[u-1][w-1] = cost
    graph[w-1][u-1] = cost

print("Enter heuristic values:")
heuristic = list(map(int, input().split()))

print("Enter source:")
src = int(input())-1

print("Enter goal:")
goal = int(input())-1
```

```python
visited = [False]*v

parent = [-1]*v

g = [float('inf')]*v

expansion_order = []

found = False
```

```python
def astar(graph, src):

    pq = []

    g[src] = 0

    heappush(pq, (heuristic[src], src))

    while pq:

        f, node = heappop(pq)

        if visited[node]:
            continue

        visited[node] = True

        expansion_order.append(node)

        if node == goal:
            return True

        for i in range(v):

            if graph[node][i] != 0:

                new_g = g[node] + graph[node][i]

                if new_g < g[i]:

                    g[i] = new_g

                    parent[i] = node

                    heappush(pq, (g[i] + heuristic[i], i))

    return False
```

```python
found = astar(graph, src)

print("\nExpansion Order:")

for x in expansion_order:
    print(x+1,end=" -> ")

print("\n\nPath to Goal:")

if found:

    path=[]

    i=goal

    while i!=-1:
        path.append(i)
        i=parent[i]

    for x in reversed(path):
        print(x+1,end=" -> ")

    print("\n\nCost =", g[goal])

else:
    print("No Path Found")
```
