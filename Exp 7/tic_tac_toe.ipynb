```python
# AI LAB EXPERIMENT - 07
# Adversarial Search: Tic-Tac-Toe using Minimax and Alpha-Beta Pruning

# ---------------------------------------------------------
# 1. Board Initialization
# ---------------------------------------------------------

board = [" " for _ in range(9)]


# ---------------------------------------------------------
# 2. Display the Board
# ---------------------------------------------------------

def print_board():
    print()
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print()


# ---------------------------------------------------------
# 3. Check Whether a Player Has Won
# ---------------------------------------------------------

def check_winner(player):

    winning_positions = [
        (0, 1, 2),  # Row 1
        (3, 4, 5),  # Row 2
        (6, 7, 8),  # Row 3
        (0, 3, 6),  # Column 1
        (1, 4, 7),  # Column 2
        (2, 5, 8),  # Column 3
        (0, 4, 8),  # Main diagonal
        (2, 4, 6)   # Other diagonal
    ]

    for a, b, c in winning_positions:

        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


# ---------------------------------------------------------
# 4. Check for Draw
# ---------------------------------------------------------

def is_draw():

    return " " not in board


# ---------------------------------------------------------
# 5. Utility / Evaluation Function
#
# MAX = X
# MIN = O
#
# X wins -> +1
# Draw   ->  0
# O wins -> -1
# ---------------------------------------------------------

def evaluate():

    if check_winner("X"):
        return 1

    if check_winner("O"):
        return -1

    return 0


# ---------------------------------------------------------
# 6. Minimax with Alpha-Beta Pruning
# ---------------------------------------------------------

def minimax_alpha_beta(depth, maximizing_player, alpha, beta):

    # Evaluate the current board
    score = evaluate()

    # Terminal state: MAX wins
    if score == 1:
        return 1

    # Terminal state: MIN wins
    if score == -1:
        return -1

    # Terminal state: Draw
    if is_draw():
        return 0


    # -----------------------------------------------------
    # MAX PLAYER - X
    # -----------------------------------------------------

    if maximizing_player:

        best_score = -float("inf")

        for i in range(9):

            # Check whether the position is empty
            if board[i] == " ":

                # Make the move
                board[i] = "X"

                # Recursively evaluate the move
                score = minimax_alpha_beta(
                    depth + 1,
                    False,
                    alpha,
                    beta
                )

                # Undo the move
                board[i] = " "

                # MAX chooses maximum value
                best_score = max(best_score, score)

                # Update Alpha
                alpha = max(alpha, best_score)

                # Alpha-Beta pruning
                if alpha >= beta:
                    break

        return best_score


    # -----------------------------------------------------
    # MIN PLAYER - O
    # -----------------------------------------------------

    else:

        best_score = float("inf")

        for i in range(9):

            # Check whether the position is empty
            if board[i] == " ":

                # Make the move
                board[i] = "O"

                # Recursively evaluate the move
                score = minimax_alpha_beta(
                    depth + 1,
                    True,
                    alpha,
                    beta
                )

                # Undo the move
                board[i] = " "

                # MIN chooses minimum value
                best_score = min(best_score, score)

                # Update Beta
                beta = min(beta, best_score)

                # Alpha-Beta pruning
                if alpha >= beta:
                    break

        return best_score


# ---------------------------------------------------------
# 7. Find the Best Move for AI
# ---------------------------------------------------------

def find_best_move():

    best_score = -float("inf")

    best_move = -1

    for i in range(9):

        # Check empty position
        if board[i] == " ":

            # Try X in this position
            board[i] = "X"

            # Evaluate this move
            score = minimax_alpha_beta(
                0,
                False,
                -float("inf"),
                float("inf")
            )

            # Undo the move
            board[i] = " "

            # Choose the move with highest score
            if score > best_score:

                best_score = score
                best_move = i

    return best_move


# ---------------------------------------------------------
# 8. Main Game
# ---------------------------------------------------------

def main():

    print("===================================")
    print("   TIC-TAC-TOE AI")
    print("   Minimax + Alpha-Beta Pruning")
    print("===================================")

    print()
    print("AI    = X")
    print("Human = O")

    print()
    print("Board positions are:")
    print()
    print(" 1 | 2 | 3")
    print("---+---+---")
    print(" 4 | 5 | 6")
    print("---+---+---")
    print(" 7 | 8 | 9")

    while True:

        # -------------------------------------------------
        # AI TURN
        # -------------------------------------------------

        print()
        print("AI is thinking...")

        move = find_best_move()

        if move != -1:

            board[move] = "X"

            print("AI selected position:", move + 1)

        print_board()


        # Check whether AI won
        if check_winner("X"):

            print("AI (X) wins!")

            break


        # Check whether game is draw
        if is_draw():

            print("The game is a draw!")

            break


        # -------------------------------------------------
        # HUMAN TURN
        # -------------------------------------------------

        while True:

            try:

                move = int(input("Enter your move (1-9): "))

                move = move - 1

                # Check valid position
                if move < 0 or move > 8:

                    print("Please enter a number from 1 to 9.")

                    continue

                # Check whether position is empty
                if board[move] != " ":

                    print("That position is already occupied.")

                    continue

                # Valid move
                board[move] = "O"

                break

            except ValueError:

                print("Please enter a valid number.")


        print_board()


        # Check whether human won
        if check_winner("O"):

            print("Human (O) wins!")

            break


        # Check whether game is draw
        if is_draw():

            print("The game is a draw!")

            break


# ---------------------------------------------------------
# 9. Start the Game
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```
