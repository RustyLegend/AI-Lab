```python
import heapq

GOAL = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

goal_position = {}

for i in range(3):
    for j in range(3):
        goal_position[GOAL[i][j]] = (i, j)
```

```python
def manhattan_distance(state):

    distance = 0

    for i in range(3):
        for j in range(3):

            value = state[i][j]

            if value != 0:
                x, y = goal_position[value]
                distance += abs(i - x) + abs(j - y)

    return distance
```

```python
def find_blank(state):

    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
```

```python
def generate_successors(state):

    row, col = find_blank(state)

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    successors = []

    for dr, dc, move in moves:

        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new_state = [list(r) for r in state]

            new_state[row][col], new_state[nr][nc] = \
                new_state[nr][nc], new_state[row][col]

            successors.append((tuple(tuple(r) for r in new_state), move))

    return successors
```

```python
def print_state(state):

    for row in state:
        print(*row)
    print()
```

```python
def greedy_best_first(start):

    open_list = []

    heapq.heappush(open_list, (manhattan_distance(start), start))

    visited = set()

    parent = {}
    move_taken = {}

    states_generated = 1
    states_expanded = 0

    while open_list:

        h, current = heapq.heappop(open_list)

        if current in visited:
            continue

        visited.add(current)
        states_expanded += 1

        if current == GOAL:

            path = []

            while current in parent:
                path.append((current, move_taken[current]))
                current = parent[current]

            path.append((current, "START"))
            path.reverse()

            print("\nSolution Found\n")

            for state, move in path:
                print("Move:", move)
                print_state(state)

            print('Goal Reached')
            print("Total Moves :", len(path) - 1)

            return

        for neighbour, move in generate_successors(current):

            if neighbour not in visited:

                heapq.heappush(
                    open_list,
                    (manhattan_distance(neighbour), neighbour)
                )

                if neighbour not in parent:
                    parent[neighbour] = current
                    move_taken[neighbour] = move

                states_generated += 1

    print("No Solution Found")
```

```python
print("Enter the initial state (use 0 for the blank):")

start = []

for _ in range(3):
    start.append(list(map(int, input().split())))

start = tuple(tuple(row) for row in start)
```

```python
greedy_best_first(start)
```
