```python
import heapq

GOAL = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)
```

```python
print("Enter the initial state (use 0 for the blank):")

start = []

for i in range(3):
    row = list(map(int, input().split()))
    start.append(row)

start = tuple(tuple(row) for row in start)
```

```python
def manhattan_distance(state):

    distance = 0

    for i in range(3):
        for j in range(3):

            if state[i][j] != 0:

                value = state[i][j]

                for x in range(3):
                    for y in range(3):
                        if GOAL[x][y] == value:
                            distance += abs(i - x) + abs(j - y)
                            break

    return distance
```

```python
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
```

```python
def generate_successors(state):

    row, col = find_blank(state)

    directions = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    successors = []

    for dr, dc, move in directions:

        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new_state = [list(r) for r in state]

            new_state[row][col], new_state[nr][nc] = \
                new_state[nr][nc], new_state[row][col]

            successors.append((tuple(tuple(r) for r in new_state), move))

    return successors
```

```python
def print_state(state):
    for row in state:
        print(*row)
    print()
```

```python
def astar(start):

    open_list = []

    heapq.heappush(
    open_list,
    (manhattan_distance(start), 0, start)
    )

    parent = {}
    move_taken = {}

    g_cost = {start: 0}

    visited = set()

    states_generated = 1
    states_expanded = 0
    max_open = 1

    while open_list:

        max_open = max(max_open, len(open_list))

        f, g, current = heapq.heappop(open_list)

        if current in visited:
            continue

        visited.add(current)
        states_expanded += 1

        if current == GOAL:

            path = []

            while current in parent:
                path.append((current, move_taken[current]))
                current = parent[current]

            path.append((current, "START"))
            path.reverse()

            print("\nSolution Found\n")

            for state, move in path:
                print("Move:", move)
                print_state(state)

            print('Goal Reached')
            print('States Generated:', states_generated)
            print('States Expanded:', states_expanded)
            print('Max Open List Size:', max_open)
            print('Total moves:', len(path)-1)

            return

        for neighbour, move in generate_successors(current):

            if neighbour in visited:
                continue

            new_g = g + 1

            if neighbour not in g_cost or new_g < g_cost[neighbour]:

                g_cost[neighbour] = new_g

                h = manhattan_distance(neighbour)

                heapq.heappush(
                    open_list,
                    (new_g + h, new_g, neighbour)
                )

                parent[neighbour] = current
                move_taken[neighbour] = move

                states_generated += 1

    print("No Solution Found")
```

```python
astar(start)
```
