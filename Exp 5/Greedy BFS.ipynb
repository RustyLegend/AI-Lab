```python
from heapq import heappush, heappop

print("----INPUT----")

v = int(input("Enter no.of vertices: "))
e = int(input("Enter no.of edges: "))

graph = [[0]*v for _ in range(v)]

print("Enter Source Destination Weight:")
for _ in range(e):
    u, w, cost = map(int, input().split())
    graph[u-1][w-1] = cost
    graph[w-1][u-1] = cost

print("Enter heuristic values:")
heuristic = list(map(int, input().split()))

print("Enter source:")
src = int(input())-1

print("Enter goal:")
goal = int(input())-1
```

```python
visited = [False]*v
parent = [-1]*v

expansion_order = []
cost = [float('inf')] * v
cost[src] = 0
found = False
```

```python
def greedy_bfs(graph, src):

    pq = []

    heappush(pq, (heuristic[src], src))

    while pq:

        h, node = heappop(pq)

        if visited[node]:
            continue

        visited[node] = True
        expansion_order.append(node)

        if node == goal:
            return True

        for i in range(v):

            if graph[node][i] != 0 and not visited[i]:

                parent[i] = node
                cost[i] = cost[node] + graph[node][i]
                heappush(pq, (heuristic[i], i))

    return False
```

```python
found = greedy_bfs(graph, src)

print("\nExpansion Order:")

for x in expansion_order:
    print(x+1, end=" -> ")

print("\n\nPath to Goal:")

if found:

    path=[]

    i=goal

    while i!=-1:
        path.append(i)
        i=parent[i]

    for x in reversed(path):
        print(x+1,end=" -> ")

    print("\n\nCost =", cost[goal])

else:
    print("No Path Found")
```
