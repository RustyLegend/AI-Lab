```python
print('----INPUT----')
v = int(input('Enter no.of vertices: '))
e = int(input('Enter no.of edges: '))

graph = [[0 for i in range(v)] for i in range(v)]

print('Source, Destination pairs: ')
for i in range(e):
  src, dest = map(int, input().split())
  graph[src-1][dest-1] = 1
  graph[dest-1][src-1] = 1

print('Enter source vertex: ')
src = int(input()) - 1

print('Enter goal vertex: ')
goal = int(input()) - 1
```

```python
visited = [False]*v
expansion_order = []
parent = [-1]*v
found = 0
```

```python
def DFS(graph, node, visited, v):
    visited[node] = True
    expansion_order.append(node)

    if node == goal:
        return True

    for i in range(v):
        if graph[node][i] != 0 and not visited[i]:
            parent[i] = node
            if DFS(graph, i, visited, v):
                return True

    return False
```

```python
found = DFS(graph, src, visited, v)

print('\n\nExpansion Order: ')
for x in expansion_order:
  print(x + 1, end = ' -> ')

print('\n\nPath to goal: ')
shortest_path = []
if found:
  i = goal
  while i != -1:
    shortest_path.append(i)
    i = parent[i]
  for x in reversed(shortest_path):
    print(x + 1, end = ' -> ')
else:
  print('No path found')
```
