```python
from collections import deque
```

```python
print('-------INITIAL STATES-------')
print('0 = Clean')
print('1 = Dirty')
room_a = int(input('Dirt in room A? (0/1): '))
room_b = int(input('Dirt in room B? (0/1): '))
room_c = int(input('Dirt in room C? (0/1): '))
room_d = int(input('Dirt in room D? (0/1): '))
A = 'A'
B = 'B'
C = 'C'
D = 'D'
pos = A
```

```python
q = deque([(pos, room_a, room_b, room_c, room_d)])
visited = set([(pos, room_a, room_b, room_c, room_d)])
parent = {(pos, room_a, room_b, room_c, room_d) : None}
ans_state = []
```

```python
while q:
    
    state = q.popleft()
    pos = state[0]
    room_a = state[1]
    room_b = state[2]
    room_c = state[3]
    room_d = state[4]
    
    if room_a == 0 and room_b == 0 and room_c == 0 and room_d == 0:
        ans_state = state
        break
        
    gen_states = []
    
    if pos == A:
        if room_a: 
            gen_states.append((pos, 0, room_b, room_c, room_d))

        gen_states.append((B, room_a, room_b, room_c, room_d))
    elif pos == B: 
        if room_b: 
            gen_states.append((pos, room_a, 0, room_c, room_d))

        gen_states.append((A, room_a, room_b, room_c, room_d))
        gen_states.append((C, room_a, room_b, room_c, room_d))
    elif pos == C: 
        if room_c: 
            gen_states.append((pos, room_a, room_b, 0, room_d))

        gen_states.append((B, room_a, room_b, room_c, room_d))
        gen_states.append((D, room_a, room_b, room_c, room_d))
    elif pos  == D:
        if room_d: 
            gen_states.append((pos, room_a, room_b, room_c, 0))

        gen_states.append((C, room_a, room_b, room_c, room_d))

    for i in gen_states: 
        if i not in visited: 
            q.append(i)
            visited.add(i)
            parent[i] = state

print('0 = Clean')
print('1 = Dirty\n')
print("Vacuum at, Room A, Room B, Room C, Room D")


if ans_state:
    path = [ans_state]
    i = ans_state
    while parent[i]:
        path.append(parent[i])
        i = parent[i]
    path.reverse()

    for i in path:
        print(i)

    print("Goal Reached")
else:
    print("No solution")
```
